```python
# Student-Specific Inputs
LAST_NAME = "Antonio"
FAVORITE_ARTIST = "RADIOHEAD"
SEED_NUM = 3

# Data Setup
customers = [f"{LAST_NAME} C{i}" for i in range(1, 6)]
products = {
    "NOTEBOOK": 50.0,
    "PEN": 20.0,
    "USB": 350.0,
    "HEADPHONES": 500.0,
    "CALCULATOR": 750.0,
}
quantities = [(SEED_NUM + ord(c)) % 5 + 1 for c in FAVORITE_ARTIST[:6]]

# Assessment Data 1: Generated Transaction Records[cite: 1, 2]
raw_transactions = [
    (customers[0], "NOTEBOOK", quantities[0]),
    (customers[1], "PEN", quantities[1]),
    (customers[2], "USB", quantities[2]),
    (customers[3], "HEADPHONES", quantities[3]),
    (customers[4], "CALCULATOR", quantities[4]),
    (customers[0], "NOTEBOOK", 2),  # Repeated purchase[cite: 1]
]

# Assessment Data 2: Organized Transaction Data[cite: 1, 2]
org_data = [
    {"customer": c, "product": p, "qty": q, "value": q * products[p]}
    for c, p, q in raw_transactions
]

# Assessment Data 3: Repeated/Distinct Transactions[cite: 1, 2]
pairs = [(t["customer"], t["product"]) for t in org_data]
repeated = list({p for p in pairs if pairs.count(p) > 1})
distinct = {
    "customers": list({t["customer"] for t in org_data}),
    "products": list({t["product"] for t in org_data}),
}

# Calculations for Summaries & Ordering
cust_sum, prod_sum = {}, {}
for t in org_data:
  cust_sum[t["customer"]] = cust_sum.get(t["customer"], 0) + t["value"]
  prod_sum[t["product"]] = prod_sum.get(t["product"], 0) + t["qty"]
ordered = sorted(org_data, key=lambda x: x["value"], reverse=True)
total_value = sum(t["value"] for t in org_data)

# Print Assessment Data (1 to 8)
print("=== ASSESSMENT DATA ===")
print("1. Generated Records:", raw_transactions)
print("2. Organized Data:", org_data)
print("3. Repeated/Distinct:", {"repeated": repeated, **distinct})
print("4. Transaction Values:", [t["value"] for t in org_data])  #[cite: 1, 2]
print("5. Customer Summaries:", cust_sum)
print("6. Product Summaries:", prod_sum)
print("7. Ordered Results:", ordered)
print("8. Overall Summary Total:", total_value)

# Final Output: Final Consolidated Transaction Report[cite: 1, 2]
print("\n" + "=" * 40)
print("FINAL CONSOLIDATED TRANSACTION REPORT")
print("=" * 40)
print(f"Total Transactions Processed: {len(raw_transactions)}")
print(f"Total Gross Revenue: ₱{total_value:.2f}")
print("\n--- Top Customer Spends ---")
for c, val in sorted(cust_sum.items(), key=lambda x: x[1], reverse=True):
  print(f"  {c}: ₱{val:.2f}")
print("\n--- Product Quantities Sold ---")
for p, q in prod_sum.items():
  print(f"  {p}: {q} units")
print("=" * 40)
```
