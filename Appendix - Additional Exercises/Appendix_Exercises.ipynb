```python
from collections import Counter, defaultdict

sys_config = {"LAST_NAME": "Antonio", "SEED_NUM": 3, "FAVORITE_ARTIST": "Radiohead"}
vals = [sys_config["SEED_NUM"] + ord(c) for c in sys_config["FAVORITE_ARTIST"]]
name = sys_config["LAST_NAME"]

# 1. Generated Performance Records[cite: 3, 4]
generated_performance_records = [
    (f"{name}_S1", "PROGRAMMING", vals[0]),
    (f"{name}_S2", "MATHEMATICS", vals[1]),
    (f"{name}_S3", "ELECTRONICS", vals[2]),
    (f"{name}_S4", "PHYSICS", vals[3]),
    (f"{name}_S5", "ENGLISH", vals[4]),
    (f"{name}_S1", "PROGRAMMING", vals[0]),  # Repeated record[cite: 3]
    (f"{name}_S2", "ELECTRONICS", None),     # Incomplete record (missing score)[cite: 3]
]
print("1. Generated Performance Records:", generated_performance_records)

# 2. Organized Student/Subject Data[cite: 3, 4]
organized_student_data = defaultdict(list)
for student, subject, score in generated_performance_records:
    organized_student_data[student].append((subject, score))
print("2. Organized Student/Subject Data:", dict(organized_student_data))

# 3. Repeated/Incomplete Records[cite: 3, 4]
counts = Counter(generated_performance_records)
repeated_incomplete_records = {
    "repeated": [item for item, count in counts.items() if count > 1],
    "incomplete": [item for item in generated_performance_records if item[2] is None]
}
print("3. Repeated/Incomplete Records:", repeated_incomplete_records)

# 4. Performance Categories[cite: 3, 4]
performance_categories = [
    (student, subject, score, "EXCELLENT" if score and score >= 110 else "GOOD" if score and score >= 100 else "PASS" if score is not None else "INCOMPLETE")
    for student, subject, score in generated_performance_records
]
print("4. Performance Categories:", performance_categories)

# 5. Student and Subject Summaries[cite: 3, 4]
valid_scores = [score for _, _, score in generated_performance_records if score is not None]
student_and_subject_summaries = {
    "min_score": min(valid_scores),
    "max_score": max(valid_scores),
    "average_score": sum(valid_scores) / len(valid_scores),
    "student_counts": dict(Counter(student for student, _, _ in generated_performance_records)),
    "subject_counts": dict(Counter(subject for _, subject, _ in generated_performance_records))
}
print("5. Student and Subject Summaries:", student_and_subject_summaries)

# 6. Ordered Results[cite: 3, 4]
ordered_results = sorted(organized_student_data.items())
print("6. Ordered Results:", ordered_results)

# Final Student Performance Report (Final Output)[cite: 3, 4]
print("\nFinal Student Performance Report:", {student: sorted(records) for student, records in ordered_results})
```

```python
from collections import Counter, defaultdict

sys_config = {"LAST_NAME": "Antonio", "SEED_NUM": 3, "FAVORITE_ARTIST": "Radiohead"}
quantities = [sys_config["SEED_NUM"] + ord(c) for c in sys_config["FAVORITE_ARTIST"]]
name = sys_config["LAST_NAME"]

# Fixed prices for the 5 products
product_prices = {
    "NOTEBOOK": 50.0,
    "PEN": 15.0,
    "USB": 250.0,
    "HEADPHONES": 500.0,
    "CALCULATOR": 350.0
}

# 1. Generated Transaction Records[cite: 6]
# Format: (Customer, Product, Quantity)
generated_transaction_records = [
    (f"{name}_C1", "NOTEBOOK", quantities[0]),
    (f"{name}_C2", "PEN", quantities[1]),
    (f"{name}_C3", "USB", quantities[2]),
    (f"{name}_C4", "HEADPHONES", quantities[3]),
    (f"{name}_C5", "CALCULATOR", quantities[4]),
    (f"{name}_C1", "NOTEBOOK", quantities[0]),  # Repeated purchase[cite: 6]
]
print("1. Generated Transaction Records:", generated_transaction_records)

# 2. Organized Transaction Data[cite: 6]
organized_transaction_data = defaultdict(list)
for customer, product, qty in generated_transaction_records:
    organized_transaction_data[customer].append((product, qty))
print("2. Organized Transaction Data:", dict(organized_transaction_data))

# 3. Repeated/Distinct Transactions[cite: 6]
counts = Counter(generated_transaction_records)
repeated_distinct_transactions = {
    "repeated": [item for item, count in counts.items() if count > 1],
    "distinct": list(counts.keys())
}
print("3. Repeated/Distinct Transactions:", repeated_distinct_transactions)

# 4. Transaction Values[cite: 6]
transaction_values = [
    (customer, product, qty, product_prices[product], qty * product_prices[product])
    for customer, product, qty in generated_transaction_records
]
print("4. Transaction Values (Customer, Product, Qty, Price, Total):", transaction_values)

# 5. Customer Summaries[cite: 6]
customer_summaries = {}
for customer, records in organized_transaction_data.items():
    total_spent = sum(qty * product_prices[prod] for prod, qty in records)
    total_items = sum(qty for _, qty in records)
    customer_summaries[customer] = {
        "total_items_bought": total_items,
        "total_amount_spent": total_spent
    }
print("5. Customer Summaries:", customer_summaries)

# 6. Product Summaries[cite: 6]
product_sales = defaultdict(lambda: {"total_qty": 0, "total_revenue": 0.0})
for _, product, qty in generated_transaction_records:
    product_sales[product]["total_qty"] += qty
    product_sales[product]["total_revenue"] += qty * product_prices[product]
product_summaries = dict(product_sales)
print("6. Product Summaries:", product_summaries)

# 7. Ordered Results[cite: 6]
ordered_results = sorted(organized_transaction_data.items())
print("7. Ordered Results:", ordered_results)

# 8. Overall Transaction Summary[cite: 6]
all_totals = [qty * product_prices[prod] for _, prod, qty in generated_transaction_records]
overall_transaction_summary = {
    "total_transactions": len(generated_transaction_records),
    "overall_revenue": sum(all_totals),
    "min_transaction_value": min(all_totals),
    "max_transaction_value": max(all_totals)
}
print("8. Overall Transaction Summary:", overall_transaction_summary)

# Final Output: Final Consolidated Transaction Report[cite: 6]
print("\nFinal Consolidated Transaction Report:", {
    "report_owner": name,
    "overall_summary": overall_transaction_summary,
    "customer_reports": customer_summaries,
    "product_reports": product_summaries,
    "ordered_records": {cust: sorted(recs) for cust, recs in ordered_results}
})
```
