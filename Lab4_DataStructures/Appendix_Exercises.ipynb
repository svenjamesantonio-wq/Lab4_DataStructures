```python
# 1. Engineering Test Data Organizer
from collections import Counter, defaultdict

# Configuration & Data Generation
sys_config = {"LAST_NAME": "Antonio", "SEED_NUM": 3, "FAVORITE_ARTIST": "Radiohead"}
vals = [sys_config["SEED_NUM"] + ord(c) for c in sys_config["FAVORITE_ARTIST"]]
name = sys_config["LAST_NAME"]

# 1. Generated Test Data[cite: 1]
generated_test_data = [
    (f"{name}_T1", "TEMPERATURE", vals[0]),
    (f"{name}_T2", "VOLTAGE", vals[1]),
    (f"{name}_T3", "CURRENT", vals[2]),
    (f"{name}_T4", "PRESSURE", vals[3]),
    (f"{name}_T5", "SPEED", vals[4]),
    (f"{name}_T1", "TEMPERATURE", vals[0]),  # Repeated record
    (f"{name}_T2", "VOLTAGE", vals[5]),      # Additional data
]

print("=== 1. GENERATED TEST DATA ===")
for idx, record in enumerate(generated_test_data, 1):
    print(f"  [{idx}] Source: {record[0]} | Category: {record[1]} | Value: {record[2]}")

# 2. Organized Test Data[cite: 1]
organized_test_data = defaultdict(list)
for src, cat, val in generated_test_data:
    organized_test_data[src].append((cat, val))

print("\n=== 2. ORGANIZED TEST DATA ===")
for src, records in organized_test_data.items():
    print(f"  * {src}:")
    for cat, val in records:
        print(f"    - Category: {cat}, Value: {val}")

# 3. Repeated/Distinct Data[cite: 1]
counts = Counter(generated_test_data)
repeated_distinct_data = {
    "repeated": [item for item, count in counts.items() if count > 1],
    "distinct": list(counts.keys())
}

print("\n=== 3. REPEATED/DISTINCT DATA ===")
print("  * Repeated Records:")
for item in repeated_distinct_data["repeated"]:
    print(f"    - {item}")
print(f"  * Total Distinct Records: {len(repeated_distinct_data['distinct'])}")

# 4. Numerical and Categorical Summaries[cite: 1]
all_vals = [v for _, _, v in generated_test_data]
numerical_and_categorical_summaries = {
    "categories_count": dict(Counter(cat for _, cat, _ in generated_test_data)),
    "min_value": min(all_vals),
    "max_value": max(all_vals),
    "average_value": round(sum(all_vals) / len(all_vals), 2)
}

print("\n=== 4. NUMERICAL AND CATEGORICAL SUMMARIES ===")
print("  * Category Counts:")
for cat, count in numerical_and_categorical_summaries["categories_count"].items():
    print(f"    - {cat}: {count} entry/entries")
print(f"  * Minimum Value: {numerical_and_categorical_summaries['min_value']}")
print(f"  * Maximum Value: {numerical_and_categorical_summaries['max_value']}")
print(f"  * Average Value: {numerical_and_categorical_summaries['average_value']}")

# 5. Processed Results[cite: 1]
processed_results = {src: sorted(records) for src, records in organized_test_data.items()}

print("\n=== 5. PROCESSED RESULTS ===")
for src, records in processed_results.items():
    print(f"  * {src} -> Sorted Records: {records}")

# 6. Ordered Results[cite: 1]
ordered_results = sorted(processed_results.items())

print("\n=== 6. ORDERED RESULTS ===")
for src, records in ordered_results:
    print(f"  * Source: {src} | Records: {records}")

# 7. Final Summary[cite: 1]
print("\n=== 7. FINAL SUMMARY ===")
for src, records in ordered_results:
    print(f"  - {src}:")
    for cat, val in records:
        print(f"      -> {cat}: {val}")
```

```python
# 2. Student Performance Data Analyzer
from collections import Counter, defaultdict

# Configuration & Data Generation
sys_config = {"LAST_NAME": "Antonio", "SEED_NUM": 3, "FAVORITE_ARTIST": "Radiohead"}
vals = [sys_config["SEED_NUM"] + ord(c) for c in sys_config["FAVORITE_ARTIST"]]
name = sys_config["LAST_NAME"]

# 1. Generated Performance Records
generated_performance_records = [
    (f"{name}_S1", "PROGRAMMING", vals[0]),
    (f"{name}_S2", "MATHEMATICS", vals[1]),
    (f"{name}_S3", "ELECTRONICS", vals[2]),
    (f"{name}_S4", "PHYSICS", vals[3]),
    (f"{name}_S5", "ENGLISH", vals[4]),
    (f"{name}_S1", "PROGRAMMING", vals[0]),  # Repeated record
    (f"{name}_S2", "ELECTRONICS", None),     # Incomplete record (missing score)
]

print("=== 1. GENERATED PERFORMANCE RECORDS ===")
for idx, (student, subject, score) in enumerate(generated_performance_records, 1):
    score_display = score if score is not None else "INCOMPLETE (None)"
    print(f"  [{idx}] Student: {student} | Subject: {subject} | Score: {score_display}")

# 2. Organized Student/Subject Data
organized_student_data = defaultdict(list)
for student, subject, score in generated_performance_records:
    organized_student_data[student].append((subject, score))

print("\n=== 2. ORGANIZED STUDENT/SUBJECT DATA ===")
for student, records in organized_student_data.items():
    print(f"  * {student}:")
    for subject, score in records:
        score_display = score if score is not None else "Missing"
        print(f"    - Subject: {subject}, Score: {score_display}")

# 3. Repeated/Incomplete Records
counts = Counter(generated_performance_records)
repeated_incomplete_records = {
    "repeated": [item for item, count in counts.items() if count > 1],
    "incomplete": [item for item in generated_performance_records if item[2] is None]
}

print("\n=== 3. REPEATED/INCOMPLETE RECORDS ===")
print("  * Repeated Records:")
for item in repeated_incomplete_records["repeated"]:
    print(f"    - {item}")
print("  * Incomplete Records:")
for item in repeated_incomplete_records["incomplete"]:
    print(f"    - {item}")

# 4. Performance Categories
performance_categories = [
    (student, subject, score, "EXCELLENT" if score and score >= 110 else "GOOD" if score and score >= 100 else "PASS" if score is not None else "INCOMPLETE")
    for student, subject, score in generated_performance_records
]

print("\n=== 4. PERFORMANCE CATEGORIES ===")
for student, subject, score, category in performance_categories:
    score_display = score if score is not None else "N/A"
    print(f"  * {student} | {subject} | Score: {score_display} -> Category: [{category}]")

# 5. Student and Subject Summaries
valid_scores = [score for _, _, score in generated_performance_records if score is not None]
student_and_subject_summaries = {
    "min_score": min(valid_scores),
    "max_score": max(valid_scores),
    "average_score": round(sum(valid_scores) / len(valid_scores), 2),
    "student_counts": dict(Counter(student for student, _, _ in generated_performance_records)),
    "subject_counts": dict(Counter(subject for _, subject, _ in generated_performance_records))
}

print("\n=== 5. STUDENT AND SUBJECT SUMMARIES ===")
print(f"  * Minimum Score: {student_and_subject_summaries['min_score']}")
print(f"  * Maximum Score: {student_and_subject_summaries['max_score']}")
print(f"  * Average Score: {student_and_subject_summaries['average_score']}")
print("  * Student Record Counts:")
for student, cnt in student_and_subject_summaries["student_counts"].items():
    print(f"    - {student}: {cnt} record(s)")
print("  * Subject Record Counts:")
for subject, cnt in student_and_subject_summaries["subject_counts"].items():
    print(f"    - {subject}: {cnt} record(s)")

# 6. Ordered Results
ordered_results = sorted(organized_student_data.items())

print("\n=== 6. ORDERED RESULTS ===")
for student, records in ordered_results:
    print(f"  * {student}: {records}")

# Final Student Performance Report (Final Output)
final_report = {student: sorted(records) for student, records in ordered_results}

print("\n=== FINAL STUDENT PERFORMANCE REPORT ===")
for student, records in final_report.items():
    print(f"  - {student}:")
    for subject, score in records:
        score_display = score if score is not None else "None"
        print(f"      -> {subject}: {score_display}")
```

```python
from collections import Counter, defaultdict

# Configuration & Student-Specific Inputs
LAST_NAME = "Antonio"
FAVORITE_ARTIST = "RADIOHEAD"
SEED_NUM = 3

# Data Setup
customers = [f"{LAST_NAME} C{i}" for i in range(1, 6)]
products = {
    "NOTEBOOK": 50.0,
    "PEN": 20.0,
    "USB": 350.0,
    "HEADPHONES": 500.0,
    "CALCULATOR": 750.0,
}
quantities = [(SEED_NUM + ord(c)) % 5 + 1 for c in FAVORITE_ARTIST[:5]]

# 1. Generated Transaction Records[cite: 5]
raw_transactions = [
    (customers[0], "NOTEBOOK", quantities[0]),
    (customers[1], "PEN", quantities[1]),
    (customers[2], "USB", quantities[2]),
    (customers[3], "HEADPHONES", quantities[3]),
    (customers[4], "CALCULATOR", quantities[4]),
    (customers[0], "NOTEBOOK", quantities[0]),  # Repeated purchase[cite: 5]
]

print("=== 1. GENERATED TRANSACTION RECORDS ===")
for idx, (cust, prod, qty) in enumerate(raw_transactions, 1):
    print(f"  [{idx}] Customer: {cust} | Product: {prod} | Quantity: {qty}")

# 2. Organized Transaction Data[cite: 5]
org_data = [
    {"customer": c, "product": p, "qty": q, "value": q * products[p]}
    for c, p, q in raw_transactions
]

print("\n=== 2. ORGANIZED TRANSACTION DATA ===")
for idx, t in enumerate(org_data, 1):
    print(f"  [{idx}] Customer: {t['customer']} -> Product: {t['product']} | Qty: {t['qty']} | Value: ₱{t['value']:.2f}")

# 3. Repeated/Distinct Transactions[cite: 5]
pairs = [(t["customer"], t["product"]) for t in org_data]
repeated = list({p for p in pairs if pairs.count(p) > 1})
distinct = {
    "customers": list({t["customer"] for t in org_data}),
    "products": list({t["product"] for t in org_data}),
}

print("\n=== 3. REPEATED/DISTINCT TRANSACTIONS ===")
print("  * Repeated Purchases:")
for rep in repeated:
    print(f"    - Customer: {rep[0]}, Product: {rep[1]}")
print("  * Distinct Customers:", distinct["customers"])
print("  * Distinct Products:", distinct["products"])

# 4. Transaction Values[cite: 5]
transaction_values = [t["value"] for t in org_data]
print("\n=== 4. TRANSACTION VALUES ===")
for idx, val in enumerate(transaction_values, 1):
    print(f"  * Transaction {idx} Value: ₱{val:.2f}")

# Calculations for Summaries & Ordering
cust_sum, prod_sum = {}, {}
for t in org_data:
    cust_sum[t["customer"]] = cust_sum.get(t["customer"], 0) + t["value"]
    prod_sum[t["product"]] = prod_sum.get(t["product"], 0) + t["qty"]

# 5. Customer Summaries[cite: 5]
print("\n=== 5. CUSTOMER SUMMARIES ===")
for c, val in cust_sum.items():
    print(f"  * {c}: Total Spend = ₱{val:.2f}")

# 6. Product Summaries[cite: 5]
print("\n=== 6. PRODUCT SUMMARIES ===")
for p, q in prod_sum.items():
    print(f"  * {p}: Total Units Sold = {q}")

# 7. Ordered Results[cite: 5]
ordered = sorted(org_data, key=lambda x: x["value"], reverse=True)
print("\n=== 7. ORDERED RESULTS (By Transaction Value) ===")
for idx, t in enumerate(ordered, 1):
    print(f"  [{idx}] {t['customer']} bought {t['qty']}x {t['product']} -> ₱{t['value']:.2f}")

# 8. Overall Transaction Summary[cite: 5]
total_value = sum(transaction_values)
print("\n=== 8. OVERALL TRANSACTION SUMMARY ===")
print(f"  * Total Transactions Count: {len(raw_transactions)}")
print(f"  * Overall Gross Revenue: ₱{total_value:.2f}")

# Final Output: Final Consolidated Transaction Report[cite: 5]
print("\n" + "=" * 50)
print("       FINAL CONSOLIDATED TRANSACTION REPORT")
print("=" * 50)
print(f"Total Transactions Processed: {len(raw_transactions)}")
print(f"Total Gross Revenue: ₱{total_value:.2f}")

print("\n--- Top Customer Spends ---")
for c, val in sorted(cust_sum.items(), key=lambda x: x[1], reverse=True):
    print(f"  • {c}: ₱{val:.2f}")

print("\n--- Product Quantities Sold ---")
for p, q in prod_sum.items():
    print(f"  • {p}: {q} unit(s)")
print("=" * 50)
```
